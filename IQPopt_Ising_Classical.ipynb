```python
import itertools
import math
```

```python
def unpack_params(params, N):
    """
    Split flat list into a and b.
    params: length N + N(N-1)/2
    """
    a = params[:N]

    b = [[0.0]*N for _ in range(N)]
    idx = N

    for i in range(N):
        for j in range(i+1, N):
            b[i][j] = params[idx]
            idx += 1

    return a, b


def energy(z, a, b):
    N = len(z)

    E = sum(a[i] * z[i] for i in range(N))
    for i in range(N):
        for j in range(i + 1, N):
            E += b[i][j] * z[i] * z[j]

    return E


def exact_minimum(params, N):
    a, b = unpack_params(params, N)

    best_E = math.inf
    best_z = None

    for z in itertools.product([-1, 1], repeat=N):
        E = energy(z, a, b)
        if E < best_E:
            best_E = E
            best_z = z

    # return best_E, best_z
    return best_E
```

```python
import json

with open("ising.json", "r") as f:
    dataset = json.load(f)

# Convert keys back to integers
dataset = {int(k): v for k, v in dataset.items()}

print("Dataset loaded successfully")
```

```python
solution = {}
num_example = 50
max_num_of_numbers = [3, 6, 9, 12, 15, 18, 21]
# max_num_of_numbers = [3, 4, 5, 6, 7, 8, 9, 10]
for n_qubits in max_num_of_numbers:
    result = []
    for j in range(num_example):
        result.append(exact_minimum(dataset[n_qubits][j], n_qubits))
        print(n_qubits, j)
    solution[n_qubits] = result
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Save dataset to disk
with open("ising_loss_exact.json", "w") as f:
    json.dump(solution, f, indent=4)

print("Dataset saved to ising_loss_exact.json")
```
