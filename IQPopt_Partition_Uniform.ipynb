```python
from Partition import *
from IQP_circuit import *
from numberpartitioning import karmarkar_karp
```

```python
import json

with open("partition.json", "r") as f:
    dataset = json.load(f)

# Convert keys back to integers
dataset = {int(k): v for k, v in dataset.items()}

print("Dataset loaded successfully")
```

```python
exp = {}
exp_ratio = {}
num_example = 50
max_num_of_numbers = [3, 6, 9, 12, 15, 18, 21]
# max_num_of_numbers = [3, 4, 5, 6, 7, 8, 9, 10]
```

```python
results = {}
results_ratio = {}

for n_qubits in max_num_of_numbers:
    result = []
    ratio = []
    for j in range(num_example):
        numbers = dataset[n_qubits][j]
        ops, coeffs = number_partition_obs(numbers)
        gates = gens(n_qubits, 'single')
        circuit = iqp.IqpSimulator(n_qubits, gates)
        params_init = np.random.uniform(0, np.pi, len(circuit.gates))
        
        trainer =  train_model(circuit = circuit, 
                       ops = ops, 
                       coeffs = coeffs, 
                       params_init = params_init, 
                       key = jax.random.PRNGKey(42), 
                       loss_fn = loss_fn, 
                       optimizer = "Adam", 
                       stepsize = 1e-3, 
                       n_iters = 1000, 
                       n_samples = 1000)

        trained_params = trainer.final_params
        result.append((trainer.losses).tolist())

        probs = circuit.probs(trained_params)
        bits = np.binary_repr(np.argmax(probs), n_qubits)
        solution = sum_selected(bits, numbers)

        solution_exact = karmarkar_karp(numbers, num_parts = 2)
        
        ratio.append(abs(solution - sum(solution_exact.partition[0])) / sum(solution_exact.partition[0]))

    results[n_qubits] = result
    results_ratio[n_qubits] = ratio

exp['single'] = results
exp_ratio['single'] = results_ratio
```
