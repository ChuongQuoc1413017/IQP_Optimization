```python
dataset = {}
num_example = 50
max_num_of_numbers = 10

for n in [3, 6, 9, 12, 15, 18, 21]:
    data = []
    count = 0
    print(n)
    while count < num_example:
        # datapoint = generate_unique_partition_instance(n, 
        #                                                min_value = 1, 
        #                                                max_value = 30)
        datapoint = np.random.randint(1, 100, n)
        result = karmarkar_karp(datapoint, num_parts = 2)
        if sum(result.partition[0]) == sum(result.partition[1]):
            data.append(datapoint.tolist())
            count += 1
            print(count, end = " ")
    print()
    dataset[n] = data
```

```python
import json

# Save dataset to disk
# data_json = [arr[i].tolist() for arr in dataset]
with open("partition.json", "w") as f:
    json.dump(dataset, f, indent=4)

print("Dataset saved to partition.json")
```

```python
dataset = []
num_example = 20
count = 0
while count < num_example:
    data = np.random.randint(1, 100, 50)
    if sum(data) % 2 == 0:
        dataset.append(data)
        count += 1
```

```python
import json

# Save dataset to disk
data_json = [arr.tolist() for arr in dataset]
with open("partition_50.json", "w") as f:
    json.dump(data_json, f, indent=4)

print("Dataset saved to partition_50.json")
```
