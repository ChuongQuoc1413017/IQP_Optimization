```python
# =====================================================
# Load data
# =====================================================

files = {

    "Classical Ising": {
        "Normal $(0,1)$": "ising_ratio_normal.json",
        r"All $\pi/4$ with perturbation": "ising_ratio_pi4.json",
        r"Uniform $(0,\pi)$": "ising_ratio_uniform.json"
    },

    "MaxCut": {
        "Normal $(0,1)$": "maxcut_ratio_normal.json",
        r"All $\pi/4$ with perturbation": "maxcut_ratio_pi4.json",
        r"Uniform $(0,\pi)$": "maxcut_ratio_uniform.json"
    },

    "Number Partition": {
        "Normal $(0,1)$": "partition_ratio_normal.json",
        r"All $\pi/4$ with perturbation": "partition_ratio_pi4.json",
        r"Uniform $(0,\pi)$": "partition_ratio_uniform.json"
    }

}

datasets = {}

for hamiltonian in files:

    datasets[hamiltonian] = {}

    for initialization, filename in files[hamiltonian].items():

        with open(filename, "r") as f:
            datasets[hamiltonian][initialization] = json.load(f)

# =====================================================
# Publication style
# =====================================================

plt.rcParams.update({
    "font.family": "serif",
    "font.size": 12,
    "axes.labelsize": 13,
    "axes.titlesize": 13,
    "legend.fontsize": 10,
    "xtick.labelsize": 11,
    "ytick.labelsize": 11,
    "figure.dpi": 300,
})

styles = {
    "full": {
        "label": "Full Connectivity",
        "marker": "o",
        "linestyle": "-"
    },
    "circular": {
        "label": "Circular Connectivity",
        "marker": "s",
        "linestyle": "--"
    },
    "single": {
        "label": "Single-Z Terms",
        "marker": "^",
        "linestyle": "-."
    }
}

hamiltonians = list(files.keys())
initializations = list(next(iter(files.values())).keys())

# =====================================================
# Create 3×3 figure
# =====================================================

fig, axes = plt.subplots(
    3,
    3,
    figsize=(12, 10),
    sharex=True,
    # sharey=True
)

for i, hamiltonian in enumerate(hamiltonians):

    for j, initialization in enumerate(initializations):

        ax = axes[i, j]

        data = datasets[hamiltonian][initialization]

        for structure in ["full", "circular", "single"]:

            qubits = sorted(map(int, data[structure].keys()))

            # Mean over 50 runs
            mean = np.array([
                np.mean(data[structure][str(q)])
                for q in qubits
            ])
            
            # Standard deviation over 50 runs
            std = np.array([
                np.std(data[structure][str(q)])
                for q in qubits
            ])

            # Use 95% CI instead
            # std = 1.96 * std / np.sqrt(50)

            ax.plot(
                qubits,
                mean,
                marker=styles[structure]["marker"],
                linestyle=styles[structure]["linestyle"],
                linewidth=2,
                markersize=5,
                label=styles[structure]["label"]
            )

            ax.fill_between(
                qubits,
                mean - std,
                mean + std,
                alpha=0.15
            )

        ax.grid(True, linestyle=":", alpha=0.6)

        ax.axhline(
            y=0,
            color="black",
            linestyle=":",
            linewidth=2,
            label='Optimal Value'
        )

        ax.set_xlim(2.6, 21.6)
        ax.set_xticks(np.arange(3, 22, 3))

        # column titles
        if i == 0:
            ax.set_title(initialization)

        # row labels
        if j == 0:
            ax.set_ylabel(
                hamiltonian + "\nRelative Approx. Ratio"
            )

        # x labels only on bottom row
        if i == 2:
            ax.set_xlabel("Number of Qubits")

# =====================================================
# Shared legend
# =====================================================

handles, labels = axes[0,0].get_legend_handles_labels()

fig.legend(
    handles,
    labels,
    loc="upper center",
    bbox_to_anchor=(0.5, 0.97),
    ncol=4,
    frameon=True
)

plt.tight_layout(rect=[0,0,1,0.95])

# plt.suptitle("Comparison of different Hamiltonians with different Initializations \n (1000 iterations, 50 problem samples, and 1000 Monte Carlo sampling for each configuration)", y=1.05)

plt.savefig(
    "IQP_all_hamiltonians.pdf",
    dpi=600,
    bbox_inches="tight"
)

plt.show()
```

```python
import pennylane as qp
import pennylane.numpy as np
from IQP_circuit import *
from MaxCut import *
from Partition import *
import jax
import json
import jax.numpy as jnp
import matplotlib.pyplot as plt
```

```python
n_qubits = 4
dev = qp.device("default.qubit", wires=n_qubits)

@qp.qnode(dev)
def circuit(params, gates):

    # |+>^n
    for i in range(n_qubits):
        qp.Hadamard(i)

    qp.Barrier()

    for theta, gate in zip(params, gates):

        term = gate[0]

        # single-Z term
        if len(term) == 1:
            qp.RZ(theta, wires=term[0])

    qp.Barrier()

    for theta, gate in zip(params, gates):

        term = gate[0]

        # Z^{n} term
        if len(term) == 2:
            qp.MultiRZ(theta, wires=term)

    qp.Barrier()

    for theta, gate in zip(params, gates):

        term = gate[0]

        # Z^{n} term
        if len(term) == 3:
            qp.MultiRZ(theta, wires=term)

    qp.Barrier()

    # back to basis
    for i in range(n_qubits):
        qp.Hadamard(i)

    return qp.state()

gates_full = gens(4, "triple")
theta = np.random.uniform(0, np.pi, len(gates_full))
# Print circuit
print(qml.draw_mpl(circuit)(theta, gates_full))
```

## Fig 1: combined panel (a)(b)(c)(d)

One large figure, three rows: **(a) General IQP** on top, **(b) Single-Z** + **(c) Circular** in the middle, **(d) Full** at the bottom. Panels (b)(c)(d) are annotated with the trainable parameters $\theta_i$ (each `RZ` / `MultiRZ` gate carries exactly one $\theta$).

To keep a **single consistent scale** across all four panels — same font size for every character, same size for every gate box and colour band — each sub-circuit is rendered at the same dpi and font sizes, then stacked as raw pixels (no per-panel rescaling). Adjust the look via the constants in the previous cell: `FS_GATE`, `FS_WIRE`, `FS_H`, `FS_TITLE`, `RENDER_DPI`, and the `GAP_H` / `GAP_V` spacing here.

Layers are colour-coded ($\alpha=0.7$): Hadamard (blue), single-Z (violet), ZZ (coral), ZZZ (green).

```python
# =====================================================
# Fig 1: helpers to render each IQP sub-circuit
# =====================================================
# All four panels are rendered with identical settings (same dpi, same font
# sizes, same shading) so that, when composed at native resolution, every
# gate box, colour band and character has exactly the same size.
import io
import pennylane as qp
import pennylane.numpy as np
import matplotlib.pyplot as plt
from IQP_circuit import gens

qml = qp  # the original circuit cells reference `qml`; alias to be safe

n_qubits = 4

plt.rcParams.update({
    "font.family": "serif",
    "figure.dpi": 300,
})

# Layer colours (same as the original Fig 1)
C_HAD = "deepskyblue"   # Hadamard
C_Z   = "violet"        # single-Z rotations
C_ZZ  = "lightcoral"    # two-qubit ZZ interactions
C_ZZZ = "lightgreen"    # three-qubit ZZZ interactions
SHADE_ALPHA = 0.7       # original shading strength

# Font sizes (identical for every panel -> consistent scale after compositing)
FS_GATE  = 24   # gate labels: RZ / MultiRZ / theta_i
FS_WIRE  = 22   # wire labels |0>
FS_H     = 22   # the Hadamard "H"
FS_TITLE = 30   # panel titles (a) (b) (c) (d)

RENDER_DPI = 400   # every panel is rasterised at this dpi -> same physical scale


def make_iqp_circuit(n_qubits, has_zz, has_zzz):
    """QNode drawing an IQP circuit with single-Z and optional ZZ / ZZZ layers."""
    dev = qp.device("default.qubit", wires=n_qubits)

    @qp.qnode(dev)
    def circuit(params, gates):
        # |+>^n
        for i in range(n_qubits):
            qp.Hadamard(i)
        qp.Barrier()

        # single-Z rotations
        for theta, gate in zip(params, gates):
            if len(gate[0]) == 1:
                qp.RZ(theta, wires=gate[0][0])

        # two-qubit ZZ interactions
        if has_zz:
            qp.Barrier()
            for theta, gate in zip(params, gates):
                if len(gate[0]) == 2:
                    qp.MultiRZ(theta, wires=gate[0])

        # three-qubit ZZZ interactions
        if has_zzz:
            qp.Barrier()
            for theta, gate in zip(params, gates):
                if len(gate[0]) == 3:
                    qp.MultiRZ(theta, wires=gate[0])

        qp.Barrier()

        # back to the computational basis
        for i in range(n_qubits):
            qp.Hadamard(i)

        return qp.state()

    return circuit


def _shade(ax, n_zz, n_zzz):
    """Shade the layer regions. Columns: H=0, single-Z=2, then ZZ, then ZZZ."""
    ax.axvspan(-0.5, 0.5, color=C_HAD, alpha=SHADE_ALPHA)   # initial Hadamards
    ax.axvspan(1.5, 2.5, color=C_Z, alpha=SHADE_ALPHA)      # single-Z layer
    col = 3
    if n_zz:
        ax.axvspan(3.5, 3.5 + n_zz, color=C_ZZ, alpha=SHADE_ALPHA)   # ZZ cols 4..3+n_zz
        col = 4 + n_zz
    if n_zzz:
        z0 = col + 1
        ax.axvspan(z0 - 0.5, z0 - 0.5 + n_zzz, color=C_ZZZ, alpha=SHADE_ALPHA)  # ZZZ
        col = z0 + n_zzz
    final_h = 4 if (not n_zz and not n_zzz) else col + 1
    ax.axvspan(final_h - 0.5, final_h + 0.5, color=C_HAD, alpha=SHADE_ALPHA)  # final Hadamards


def _relabel(ax, theta_labels):
    """Turn wire labels into |0>, size the H gates, and rewrite RZ/MultiRZ."""
    for t in ax.texts:
        s = t.get_text()
        if s in ["0", "1", "2", "3"]:
            t.set_text(r"$|0\rangle$")
            t.set_fontsize(FS_WIRE)
        elif s == "H":
            t.set_fontsize(FS_H)

    gate_texts = [t for t in ax.texts if t.get_text() in ["RZ", "MultiRZ"]]
    if theta_labels:
        # index parameters left-to-right (by column), then top-to-bottom
        gate_texts.sort(key=lambda t: (round(t.get_position()[0], 3),
                                       t.get_position()[1]))
        for i, t in enumerate(gate_texts, start=1):
            t.set_text(rf"$\theta_{{{i}}}$")
            t.set_fontsize(FS_GATE)
    else:
        for t in gate_texts:
            t.set_fontsize(FS_GATE)


def render_panel(mode, theta_labels, title):
    """Draw one IQP circuit (with its title) and return it as an RGBA image."""
    gates = gens(n_qubits, mode)
    n_zz = sum(1 for g in gates if len(g[0]) == 2)
    n_zzz = sum(1 for g in gates if len(g[0]) == 3)

    theta = np.random.uniform(0, np.pi, len(gates))

    circuit = make_iqp_circuit(n_qubits, has_zz=n_zz > 0, has_zzz=n_zzz > 0)
    fig, ax = qml.draw_mpl(circuit, decimals=None)(theta, gates)

    _shade(ax, n_zz, n_zzz)
    _relabel(ax, theta_labels)
    ax.set_title(title, fontsize=FS_TITLE, fontfamily="serif")

    buf = io.BytesIO()
    fig.savefig(buf, format="png", dpi=RENDER_DPI, bbox_inches="tight")
    plt.close(fig)
    buf.seek(0)
    return plt.imread(buf)
```

```python
# =====================================================
# Assemble the combined Fig 1 panel at NATIVE scale
#   (a) general IQP          -> row 1 (full width)
#   (b) single  (c) circular -> row 2
#   (d) full                 -> row 3
# Panels are stacked as raw pixels (no per-panel rescaling), so every gate
# box, colour band and character keeps exactly the same size across a/b/c/d.
# =====================================================
import numpy as onp

# titles are baked into each panel so their text matches the circuit scale
img_a = render_panel("triple",   theta_labels=False, title="(a) General IQP  (single-Z + ZZ + ZZZ)")
img_b = render_panel("single",   theta_labels=True,  title="(b) Single-Z terms")
img_c = render_panel("circular", theta_labels=True,  title="(c) Circular connectivity")
img_d = render_panel("full",     theta_labels=True,  title="(d) Full connectivity")


def _pad_height(img, height):
    """White-pad an image at the bottom to reach `height`."""
    h, w = img.shape[:2]
    if h >= height:
        return img
    return onp.concatenate([img, onp.ones((height - h, w, 4))], axis=0)


def _pad_width(img, width):
    """White-pad an image left/right to centre it in `width`."""
    h, w = img.shape[:2]
    if w >= width:
        return img
    total = width - w
    left = total // 2
    right = total - left
    return onp.concatenate([onp.ones((h, left, 4)), img, onp.ones((h, right, 4))], axis=1)


# all panels share the same pixel height (4 wires + one-line title)
H = max(im.shape[0] for im in (img_a, img_b, img_c, img_d))
img_a, img_b, img_c, img_d = (_pad_height(im, H) for im in (img_a, img_b, img_c, img_d))

GAP_H = int(0.06 * H)   # horizontal gap between (b) and (c)
GAP_V = int(0.08 * H)   # vertical gap between rows

row1 = img_a
row2 = onp.concatenate([img_b, onp.ones((H, GAP_H, 4)), img_c], axis=1)
row3 = img_d

W = max(row1.shape[1], row2.shape[1], row3.shape[1])
row1, row2, row3 = (_pad_width(r, W) for r in (row1, row2, row3))

vgap = onp.ones((GAP_V, W, 4))
canvas = onp.concatenate([row1, vgap, row2, vgap, row3], axis=0)

# display / save at native pixels (single imshow -> uniform scale everywhere)
DPI = 200
fig, ax = plt.subplots(figsize=(W / DPI, canvas.shape[0] / DPI), dpi=DPI)
ax.imshow(canvas)
ax.axis("off")
fig.subplots_adjust(left=0, right=1, top=1, bottom=0)

fig.savefig("iqp_circuits_panel.pdf", dpi=DPI, bbox_inches="tight", pad_inches=0.05)
fig.savefig("iqp_circuits_panel.png", dpi=DPI, bbox_inches="tight", pad_inches=0.05)
plt.show()
```
