```python
from IQP_circuit import *
```

```python
def obs(n_qubits: int, corr: int = 1) -> np.array:
    
    obs = []
    
    for i in range(n_qubits):
        ob = [0]*n_qubits
        ob[i] = 1
        obs.append(ob)
        
    if corr == 2:
        for i in range(n_qubits):
            for j in range(i+1, n_qubits):
                ob = [0]*n_qubits
                ob[i] = 1
                ob[j] = 1
                obs.append(ob)
                
    return np.array(obs)
```

```python
import json

with open("ising.json", "r") as f:
    dataset = json.load(f)

# Convert keys back to integers
dataset = {int(k): v for k, v in dataset.items()}

print("Dataset loaded successfully")
```

```python
with open("ising_loss_exact.json", "r") as f:
    exact = json.load(f)

# Convert keys back to integers
exact = {int(k): v for k, v in exact.items()}

print("Dataset loaded successfully")
```

```python
exp = {}
exp_ratio = {}
num_example = 50
max_num_of_numbers = [3, 6, 9, 12, 15, 18, 21]
# max_num_of_numbers = [3, 4, 5, 6, 7, 8, 9, 10]
```

```python
results = {}
results_ratio = {}

for n_qubits in max_num_of_numbers:
    result = []
    ratio = []
    for j in range(num_example):
        coeffs = np.array(dataset[n_qubits][j])
        ops = obs(n_qubits, 2)
        gates = gens(n_qubits, 'single')
        circuit = iqp.IqpSimulator(n_qubits, gates)
        params_init = np.ones(len(circuit.gates))*np.pi/4 + 0.05*np.random.normal(0, 1, len(circuit.gates))
        
        trainer =  train_model(circuit = circuit, 
                       ops = ops, 
                       coeffs = coeffs, 
                       params_init = params_init, 
                       key = jax.random.PRNGKey(42), 
                       loss_fn = loss_fn, 
                       optimizer = "Adam", 
                       stepsize = 1e-3, 
                       n_iters = 1000, 
                       n_samples = 1000)

        trained_params = trainer.final_params
        result.append((trainer.losses).tolist())
        ratio.append(float((exact[n_qubits][j] - trainer.losses[-1]) / exact[n_qubits][j]))

    results[n_qubits] = result
    results_ratio[n_qubits] = ratio

exp['single'] = results
exp_ratio['single'] = results_ratio
```
