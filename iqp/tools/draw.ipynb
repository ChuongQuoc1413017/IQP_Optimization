```python
import pennylane as qml
import numpy as np
from iqp.core.ansatz import apply_ansatz, n_params_for, init_params
from iqp.core.problems import ising_obs

def draw_ansatz(ansatz, spec, n=4):
    ops = ising_obs(n, 2)
    coeffs = np.ones(len(ops))          
    p = init_params('normal', n_params_for(ansatz, spec, n), n)
    dev = qml.device('default.qubit', wires=n)

    @qml.qnode(dev)
    def circ(params):
        apply_ansatz(params, ops, coeffs, n, ansatz, spec)
        return qml.expval(qml.PauliZ(0))

    fig, ax = qml.draw_mpl(circ)(p)
    ax.set_title(f'{ansatz}/{spec}  (n={n})')
    return fig
```

```python
for ansatz, spec in [('iqp','single'), ('iqp','circular'), ('iqp','full'),
                     ('hea', 2), ('qaoa', 2)]:
    draw_ansatz(ansatz, spec, n=4)
```
